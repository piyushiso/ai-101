## TASK 5: Iteration Using for Loops

**Concepts:** Iteration, Accumulators, Loop Control

Create a marks analyzer:

- Input marks for n subjects
- Output total, average, highest, lowest

```python
# Task #5

n = 3
total = 0
highest = float('-inf')
lowest = float('inf')
for i in range(1, n+1):
    mark = int(input(f"Enter mark of subject #{i}"))
    total += mark
    highest = max(highest, mark)
    lowest = min(lowest, mark)
    
print(f"Total Marks:\t{total}\nHighest:\t{highest}\nLowest:\t\t{lowest}\nAverage:\t{round(total/n, 2)}")
```

## TASK 7: String Processing & Traversal

**Concepts:** Strings, Indexing, Iteration

Analyze a user-entered sentence:

- Count vowels, consonants, digits, spaces
- Convert case and remove extra spaces

Reasoning Questions

1. Why is string processing critical in cybersecurity and NLP?
2. How can improper string handling introduce vulnerabilities?

```python
# Task #7

sentence = input("Enter a sentence: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0
extras = 0

vowel_string = "aeiou"
consonant_string = "bcdfghjklmnpqrstvwxyz"
digit_string = "0123456789"

for ch in sentence.lower():
    if ch in vowel_string:
        vowels += 1
    elif ch in consonant_string:
        consonants += 1
    elif ch in digit_string:
        digits += 1
    elif ch == ' ':
        spaces += 1
    else:
        extras += 1

print(f"Sentence: {sentence}")
print(f"Length: {len(sentence)} | Vowels: {vowels} | Consonants: {consonants} | Digits: {digits} | Spaces: {spaces} | Others: {extras}")
print(f"Lower Case & Stripping Extras Spaces: {sentence.lower().strip()}")
        
```

## TASK 8: Functions

**Concepts:** Functions, Parameters, Return Values, Reusability

The main program should call these functions, not duplicate logic.

**Conceptual / Brainstorming Questions**

```python
# Task #8

def avg(total, n):
    return round(total/n, 2)

def grade(marks):
    '''

        A+: 95 - 100
        A : 90 - 94
        B+: 80 - 89
        B : 70 - 79
        C+: 60 - 69
        C : 50 - 59
        F : 0 - 49
        Else Invalid
        PS: Both extreme values are inclusive.
    '''
    marks = int(marks)
    if marks < 0 or marks > 100:
        return "Invalid"
    elif marks >= 95:
        return "A+"
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B+"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C+"
    elif marks >= 50:
        return "C"
    else:
        return "F"
    
def display(total, avg, grade):
    print(f"Total:\t\t{total}\nAverage:\t{avg}\nGrade:\t\t{grade}")
    
def main():
    total = 480
    n = 5
    
    average = avg(total, n)
    
    display(total, average, grade(average))
    
if __name__ == "__main__":
    main()
    
```
